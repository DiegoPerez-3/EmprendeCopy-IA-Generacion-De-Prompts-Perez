## 6.2 Configuración de Groq

La API key se lee desde un Secret de Google Colab llamado `GROQ_API_KEY`. Si la notebook se ejecuta de forma local, también puede leerse desde una variable de entorno con el mismo nombre.

De esta manera la clave no queda escrita en el código ni se publica en GitHub.


```python
import os
import json
from groq import Groq

# Primero intento leer la key desde Colab; si no estoy en Colab, busco una variable de entorno.
api_key = None

try:
    from google.colab import userdata
    api_key = userdata.get("GROQ_API_KEY")
except Exception:
    api_key = os.environ.get("GROQ_API_KEY")

if not api_key:
    raise RuntimeError(
        "No se encontró GROQ_API_KEY. "
        "En Colab agregala en Secrets; en local definila como variable de entorno."
    )

# Uso el mismo modelo que se trabajó en el material del curso.
MODELO = "openai/gpt-oss-20b"

client = Groq(api_key=api_key)

print("Groq listo.")
print("Modelo:", MODELO)
```

## 6.3 Funciones para consultar el modelo y estimar costos

Para no repetir código armé una función que hace la consulta y devuelve tanto la respuesta como los tokens utilizados. Con esos datos hago una estimación simple del costo de cada prueba.


```python
# Precios que usé para estimar el costo de las pruebas.
# Los consulté en la documentación de Groq el 23/08/2026.
PRECIO_INPUT_USD_M = 0.075
PRECIO_OUTPUT_USD_M = 0.30

def estimar_costo(usage):
    """Estima cuánto costó una consulta usando los tokens informados por Groq."""
    input_tokens = getattr(usage, "prompt_tokens", 0) or 0
    output_tokens = getattr(usage, "completion_tokens", 0) or 0

    costo_input = (input_tokens / 1_000_000) * PRECIO_INPUT_USD_M
    costo_output = (output_tokens / 1_000_000) * PRECIO_OUTPUT_USD_M

    return {
        "input_tokens": input_tokens,
        "output_tokens": output_tokens,
        "total_tokens": input_tokens + output_tokens,
        "costo_estimado_usd": costo_input + costo_output,
    }


def chat(user, system="Respondé en español, de forma clara y breve.", temperature=0):
    """Hace una consulta a Groq y devuelve la respuesta junto con los datos de uso."""
    respuesta = client.chat.completions.create(
        model=MODELO,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": user},
        ],
        temperature=temperature,
        reasoning_effort="low",
        max_completion_tokens=1200,
    )

    texto = respuesta.choices[0].message.content
    metricas = estimar_costo(respuesta.usage)

    return texto, metricas
```

# 7. Fast Prompting: Zero-shot vs Few-shot

## 7.1 Primera prueba: zero-shot

Primero le doy al modelo la tarea y los datos sin mostrarle ejemplos previos. Esta respuesta funciona como punto de comparación.


```python
prompt_zero_shot = f"""
Creá una publicación promocional para {caso['plataforma']} para una {caso['tipo_negocio']}.
Servicio: {caso['producto_servicio']}.
Promoción: {caso['promocion']}.
Público objetivo: {caso['publico']}.
Tono: {caso['tono']}.
Objetivo: {caso['objetivo']}.

Incluí un título, un copy breve, una llamada a la acción y hashtags.
"""

respuesta_zero, metricas_zero = chat(prompt_zero_shot, temperature=0)

print("=== RESPUESTA ZERO-SHOT ===")
print(respuesta_zero)
print("\n=== USO ===")
print(metricas_zero)
```

### Observación de la primera prueba

El contenido es entendible y sirve como publicación, pero el modelo toma varias decisiones por su cuenta: cambia el formato, genera más hashtags de los que necesito y agrega un enlace en la bio/DM que no estaba dentro de los datos originales.


## 7.2 Segunda prueba: few-shot

En esta prueba agrego ejemplos completos antes del caso nuevo. La idea es mostrarle al modelo no solo qué tiene que hacer, sino también qué estructura quiero repetir.


```python
prompt_few_shot = f"""
Sos especialista en copywriting para pequeños negocios.

Generá contenido promocional y respondé EXACTAMENTE con esta estructura:

Título: ...
Copy: ...
CTA: ...
Hashtags: #... #... #... #... #...

Reglas:
- El copy debe tener como máximo 80 palabras.
- Debe haber exactamente 5 hashtags.
- No inventes precios, condiciones ni características no proporcionadas.
- No agregues explicaciones antes ni después.

EJEMPLO 1
Negocio: Barbería
Servicio: Corte + barba
Promoción: 15 % de descuento los martes
Público: Hombres de 18 a 40 años
Plataforma: Instagram
Tono: Cercano
Objetivo: Reservar turnos

Título: Martes de estilo
Copy: Renová tu look con corte y barba y aprovechá un 15 % de descuento todos los martes. Una propuesta simple para mantener tu estilo con atención profesional.
CTA: Reservá tu turno para este martes.
Hashtags: #Barbería #Corte #Barba #Estilo #Promoción

EJEMPLO 2
Negocio: Cafetería
Servicio: Combo de café y medialunas
Promoción: Precio especial de 8 a 11 h
Público: Personas que trabajan o estudian en la zona
Plataforma: Instagram
Tono: Cálido
Objetivo: Aumentar visitas por la mañana

Título: Empezá la mañana con algo rico
Copy: Hacé una pausa y disfrutá nuestro combo de café y medialunas con precio especial de 8 a 11 h. Ideal para arrancar el día antes del trabajo o el estudio.
CTA: Pasá esta mañana y disfrutá el combo.
Hashtags: #Café #Desayuno #Medialunas #Cafetería #BuenDía

CASO NUEVO
Negocio: {caso['tipo_negocio']}
Servicio: {caso['producto_servicio']}
Promoción: {caso['promocion']}
Público: {caso['publico']}
Plataforma: {caso['plataforma']}
Tono: {caso['tono']}
Objetivo: {caso['objetivo']}
"""

respuesta_few, metricas_few = chat(prompt_few_shot, temperature=0)

print("=== RESPUESTA FEW-SHOT ===")
print(respuesta_few)
print("\n=== USO ===")
print(metricas_few)
```

## 7.3 Evaluación automática del formato

Para comparar las dos respuestas verifico cosas concretas: si aparecen las secciones esperadas y cuántos hashtags devuelve cada salida. Esta evaluación se hace con Python y no agrega otra llamada a la API.


```python
import re

def evaluar_respuesta(texto):
    hashtags = re.findall(r"(?<!\w)#[\wÁÉÍÓÚÜÑáéíóúüñ]+", texto)

    return {
        "tiene_titulo": "título:" in texto.lower() or "titulo:" in texto.lower(),
        "tiene_copy": "copy:" in texto.lower(),
        "tiene_cta": "cta:" in texto.lower(),
        "tiene_hashtags": "hashtags:" in texto.lower(),
        "cantidad_hashtags_detectados": len(hashtags),
        "cumple_5_hashtags": len(hashtags) == 5,
    }

evaluacion_zero = evaluar_respuesta(respuesta_zero)
evaluacion_few = evaluar_respuesta(respuesta_few)

print("ZERO-SHOT")
print(json.dumps(evaluacion_zero, indent=2, ensure_ascii=False))

print("\nFEW-SHOT")
print(json.dumps(evaluacion_few, indent=2, ensure_ascii=False))
```

### Escala de evaluación (0 a 10)

Además de la comprobación automática, uso una escala simple con cinco criterios de 2 puntos cada uno:

| Criterio | Máximo | Zero-shot | Few-shot |
|---|---:|---:|---:|
| Respeta la estructura solicitada | 2 | 0 | 2 |
| Devuelve exactamente 5 hashtags | 2 | 0 | 2 |
| El copy tiene como máximo 80 palabras | 2 | 2 | 2 |
| Incluye una CTA clara y accionable | 2 | 2 | 2 |
| No agrega canales, condiciones o datos no proporcionados | 2 | 0 | 2 |
| **Total** | **10** | **4/10** | **10/10** |

La diferencia principal no está en decidir cuál texto es “más creativo”, sino en medir cuál se acerca mejor a las reglas que necesito para EmprendeCopy.

Con esta escala, el resultado pasa de **4/10 con zero-shot a 10/10 con few-shot**.


### Qué me deja esta comparación

Zero-shot es útil para obtener una respuesta rápida con pocos tokens de entrada. Few-shot, en cambio, usa un prompt más largo pero me da más control sobre el formato.

Para este proyecto prefiero esa mayor consistencia porque la salida tiene que ser reutilizable con distintos negocios.

# 9. Versión final optimizada

## 9.1 Copy + prompt visual en una sola consulta

Las pruebas anteriores se ejecutan por separado porque forman parte del experimento.

En un uso real no tendría sentido ejecutar zero-shot y few-shot cada vez. Por eso la versión final de EmprendeCopy hace una sola consulta a Groq y devuelve:

- título;
- copy;
- CTA;
- cinco hashtags;
- prompt visual.

La salida se organiza con **Structured Outputs / JSON Schema**.


```python
ESQUEMA_EMPRENDECOPY = {
    "type": "json_schema",
    "json_schema": {
        "name": "emprendecopy_output",
        "strict": True,
        "schema": {
            "type": "object",
            "properties": {
                "titulo": {"type": "string"},
                "copy": {"type": "string"},
                "cta": {"type": "string"},
                "hashtags": {
                    "type": "array",
                    "items": {"type": "string"}
                },
                "image_prompt": {"type": "string"}
            },
            "required": ["titulo", "copy", "cta", "hashtags", "image_prompt"],
            "additionalProperties": False
        }
    }
}


def generar_emprendecopy(datos):
    """
    Genera el contenido de EmprendeCopy en una sola llamada
    y devuelve también los datos de uso de la API.
    """
    instrucciones = """
Sos especialista en copywriting y prompts visuales para pequeños negocios.

A partir de los datos recibidos:
1. Creá un título breve.
2. Creá un copy de máximo 80 palabras.
3. Creá una CTA concreta.
4. Devolvé exactamente 5 hashtags.
5. Creá un image_prompt en inglés, de máximo 70 palabras, para una imagen publicitaria.

El image_prompt debe definir sujeto, entorno, estilo, encuadre, iluminación,
paleta y composición. Debe dejar espacio limpio para agregar copy posteriormente.
No debe pedir letras, palabras, tipografía, logos, carteles, marcas ni texto legible.

No inventes precios, descuentos, fechas, servicios ni características no presentes
en los datos del usuario.
"""

    respuesta = client.chat.completions.create(
        model=MODELO,
        messages=[
            {"role": "system", "content": instrucciones},
            {
                "role": "user",
                "content": json.dumps(datos, ensure_ascii=False)
            },
        ],
        reasoning_effort="low",
        max_completion_tokens=1200,
        response_format=ESQUEMA_EMPRENDECOPY,
    )

    resultado = json.loads(respuesta.choices[0].message.content)
    metricas = estimar_costo(respuesta.usage)

    # Como control extra, reviso que hayan vuelto exactamente cinco hashtags.
    if len(resultado.get("hashtags", [])) != 5:
        print(
            "Advertencia: se esperaban 5 hashtags y se recibieron",
            len(resultado.get("hashtags", []))
        )

    return resultado, metricas
```

## 9.2 Ejecución de la versión final

A continuación uso la versión optimizada con el mismo caso de la peluquería. Este es el flujo que utilizaría normalmente EmprendeCopy.


```python
resultado_final, metricas_final = generar_emprendecopy(caso)

print("=== RESULTADO FINAL ===")
print(json.dumps(resultado_final, indent=2, ensure_ascii=False))

print("\n=== USO Y COSTO APROXIMADO ===")
print(json.dumps(metricas_final, indent=2, ensure_ascii=False))
```

# 11. Optimización de consultas y costos

Durante la notebook hago tres consultas porque necesito mostrar el experimento:

- 1 para zero-shot;
- 1 para few-shot;
- 1 para la versión final.

Las dos primeras sirven para comparar técnicas. En el uso normal del proyecto solamente necesito la versión final:

**Datos del negocio → 1 consulta a Groq → contenido promocional + prompt visual**

La generación de la imagen se realiza manualmente en ChatGPT, por lo que no suma una llamada automatizada a una API de imágenes.


```python
def mostrar_resumen_costos():
    filas = [
        ("Zero-shot (experimento)", metricas_zero),
        ("Few-shot (experimento)", metricas_few),
        ("Versión final (uso real)", metricas_final),
    ]

    print(f"{'Etapa':30} {'Tokens':>10} {'Costo USD aprox.':>18}")
    print("-" * 62)

    for nombre, m in filas:
        print(
            f"{nombre:30} "
            f"{m['total_tokens']:>10} "
            f"{m['costo_estimado_usd']:>18.8f}"
        )

mostrar_resumen_costos()
```

# 12. Limitaciones

Aunque la POC llega al resultado esperado, encontré algunas limitaciones que tendría en cuenta si el proyecto siguiera creciendo:

- Los modelos generativos pueden producir respuestas diferentes aunque reciban instrucciones similares, por lo que no puedo garantizar una salida idéntica en cada ejecución.
- El contenido generado necesita una revisión humana antes de publicarse para comprobar que no aparezca información que el usuario no haya proporcionado.
- En la parte visual comprobé que una restricción puede no cumplirse completamente. En la primera imagen apareció texto no solicitado y fue necesario refinar el prompt.
- La generación de la imagen se hace de forma manual en esta POC, por lo que el flujo no está automatizado de principio a fin.
- Few-shot mejora la consistencia para este caso, pero utiliza más tokens de entrada que zero-shot.

Estas limitaciones no impiden que la solución funcione como prueba de concepto, pero marcan aspectos que habría que tener en cuenta en una versión más completa.

# 13. Posibles mejoras

Una mejora que podría probar más adelante es el **encadenado de prompts**.

Por ejemplo, después de generar el contenido podría ejecutar un segundo prompt que actúe como revisor. Ese segundo paso recibiría la salida anterior y comprobaría si respeta el formato, si la cantidad de hashtags es correcta y si se agregó información que no estaba en los datos originales.

No lo incorporé en esta POC porque implicaría una segunda consulta a la API por cada publicación. Uno de los objetivos del proyecto fue justamente reducir llamadas innecesarias, por eso preferí mantener una sola consulta en el flujo final y dejar la revisión humana como último control.

Si el proyecto creciera y esa validación automática justificara el costo extra, el encadenado sería una de las primeras mejoras que probaría.
